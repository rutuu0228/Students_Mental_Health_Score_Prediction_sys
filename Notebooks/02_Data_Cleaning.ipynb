# Data Cleaning :

### Objective:



- Load raw dataset
- Remove duplicates
- Handle missing values
- Fix inconsistent values
- Treat outliers (if required)
- Save cleaned dataset

```python
# Import Required Libraries for cleaning 

import numpy as np
import pandas as pd
```

```python
# Load The Raw Dataset
df = pd.read_csv("../Data/01_Raw/Student Social Media And Mental Health Impact.csv")
```

```python
# Verify the dataset has been loaded successfully
df.head()
```

### Check Missing Values:

```python
df.isnull().sum()
```

### Check Duplicate Records:

```python
# Check the total number of duplicate rows present in the dataset
df.duplicated().sum()
```

### Removing Duplicate Values:

```python
# Remove duplicate rows from the dataset
df.drop_duplicates(inplace=True)
```

```python
# Check the number of duplicate rows after removal
df.duplicated().sum()
```

### Observation

- **No missing values** values are present in the dataset.
- The dataset contained **2 duplicate records**.
- Duplicate rows were removed to avoid repeated data during model training.
- After removal, the dataset contains **no duplicate values**.

### Identify Categorical and Numerical Columns:

```python
# Identify categorical columns in the dataset
categorical_columns = df.select_dtypes(include='str').columns

categorical_columns.to_list()
```

```python
# Identify numerical columns in the dataset
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

numerical_columns.to_list()
```

### Check Unique Values in Categorical Columns 

(Check the Inconsistencies in categorical column)

```python
# Display unique values of each categorical column to identify inconsistent entries

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

### Observation

- Categorical columns were checked for inconsistent values.
- No spelling errors, duplicate categories, or formatting issues were found.
- Categorical data is already clean and ready for further preprocessing.

### Check Numerical Column Ranges

```python
# Display minimum and maximum values of numerical columns to identify unusual values

for column in numerical_columns:
    print(f"\n{column}:")
    print("Minimum value:", df[column].min())
    print("Maximum value:", df[column].max())
```

### Observation

- Numerical columns were checked for invalid or unusual values.
- Most numerical features contain values within acceptable ranges.
- A negative value was found in **Physical_Activity_Hours (-0.4)**, which is not logically valid.
- This inconsistency needs to be handled during the data cleaning process.

### Handle Invalid Numerical Values:

```python
# Check the minimum physical activity value
df['Physical_Activity_Hours'].min()
```

```python
# Replace negative values in Physical_Activity_Hours with 0
df.loc[df['Physical_Activity_Hours'] < 0, 'Physical_Activity_Hours'] = 0

# Explanation of the code above:

#df['Physical_Activity_Hours'] < 0 → Finds all rows where the value is negative.
#df.loc[...] → Selects those rows.
# = 0 → Replaces the negative values with 0.
```

### Verify the Cleaned Dataset:

```python
# Check the dataset after cleaning 

# check missing values:
df.isnull().sum()
```

```python
# check duplicates:
df.duplicated().sum()
```

```python
# check minimum value of Physical_Activity_Hours:
df['Physical_Activity_Hours'].min()
```

### Save the Cleaned Dataset

```python
# Save the cleaned dataset to a new CSV file for further analysis
df.to_csv("../Data/Cleaned/Student Social Media And Mental Health Impact Cleaned Data.csv",index=False)
```

## Conclusion

- The dataset have been successfully cleaned by removing duplicate records.
- No missing values are present in the dataset.
- Invalid numerical values were identified and corrected.
- Categorical data was consistent and required no modifications.
- The cleaned dataset is now ready for Exploratory Data Analysis (EDA) and feature engineering.
