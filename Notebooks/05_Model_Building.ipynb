# Model Building

### Objective

- Prepare the processed data for machine learning.
- Separate input features (`X`) and target (`y`).
- Split the data into training and testing sets.
- Train multiple regression models.
- Compare their performance and select the best model.

## Import libraries

```python
# Import required libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
```

## Load the processed dataset

```python
# Load the final processed dataset
df = pd.read_csv("../Data/03_Processed/mental_health_processed.csv")

# Display the shape and first few rows
print("Dataset Shape:", df.shape)
df.head()
```

### Key Insights

- The processed dataset contains the features prepared during feature engineering.
- `Mental_Health_Score` is the target variable.
- The dataset is ready for machine learning model development.

## Separate Features and Target

- `X` contains the input features used by the model.
- `y` contains the value we want the model to predict.
- Here, `Mental_Health_Score` is the target.

```python
# Separate input features and target varibales

X = df.drop(columns=["Mental_Health_Score"])
y = df["Mental_Health_Score"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)
```

## Train-Test Split

- The dataset is divided into two parts:
  - **Training data** → used to teach the model.
  - **Testing data** → used to check how well the model performs on new data.
- **80%** of the data is used for training.
- **20%** of the data is used for testing.

```python
# Split the data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20, # This Keep 20% of the data for testing
    random_state=42
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)
print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

### Key Insights

- 80% of the data is used for training the models.
- 20% is kept aside for testing.
- The test data is not used during model training.
- This helps us check how well the model works on unseen student data.

```python
# Import regression models
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.tree import DecisionTreeRegressor
```

## Create Regression Models

- Four regression models are selected for comparison.
   1. Linear Regression
   2. Decision Tree
   3. Random Forest
   4. Gradiant Boosting
- Each model will learn the relationship between the student features and `Mental_Health_Score`.
- Their performance will be compared after training.

# 1. Linear Regression Model

- Linear Regression is a simple model 
- It tries to find the relationship between the input features and target variable.
- It is used as a baseline model for comparison with other algorithms

### Linear Regression Model - Creation

```python
# Import Linear Regression
from sklearn.linear_model import LinearRegression

# Create the Linear Regression model
linear_model = LinearRegression()

# Train the Linear Regression model
linear_model.fit(X_train,y_train)# .fit() = teach the model using the training data


print("Linear Regression Model Is Trained Successfully!")
```

### Linear Regression - Prediction

```python
# Make predictions using the trained Linear Regression model
y_pred_linear = linear_model.predict(X_test)

# Display the first 10 predictions
y_pred_linear[:10]
```

- The trained model is used to predict 'Mental_Health_Score' for the test data
- 'X_test' contains data that was not used during training
- The predicted scores are stored in 'y_pred_linear'

### Linear Regression - Model Evaluation

```python
# Import evaluation metrics 
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Calculate evaluation metrics for Linear Regression

linear_mae = mean_absolute_error(y_test,  y_pred_linear)

linear_rmse = np.sqrt(mean_squared_error(y_test, y_pred_linear))

linear_r2 = r2_score(y_test,y_pred_linear)
```

```python
# Store the evaluation metrics for Linear Regression

linear_results = pd.DataFrame({
    "Model": ["Linear Regression"],
    "MAE" :[linear_mae],
    "RMSE":[linear_rmse],
    "R2 Score": [linear_r2]
})

linear_results
```

### Key Insights

- Linear Regression achieved an **MAE of 0.4758**.
- The **RMSE is 0.6135**.
- The **R² Score is 0.7892**, meaning the model explains a substantial portion of the variation in `Mental_Health_Score`.
- This result will be used as a baseline for comparison with the other regression models.

# 2. Decision Tree

- Decision Tree Regression predicts the target by making a series of decisions based on the input features
- It can capture non-linear relationships between features and 'Mental_Health_Score'
- The data is divided into smaller groups until the model reaches a prediction 

### Decision Tree model - Creation

```python
# Import Decision Tree Regressor 
from sklearn.tree import DecisionTreeRegressor

# Create the Decision Tree model
decision_tree_model = DecisionTreeRegressor(random_state= 42)

# Train the Decision Tree Model
decision_tree_model.fit(X_train, y_train)

print("Decision Tree Trained Successfully!")
```

### Decision Tree - Prediction

```python
# Make prediciton using the trained Decision Tree model
y_pred_tree = decision_tree_model.predict(X_test)

# Display the first 10 predictions
y_pred_tree[:10]
```

- The trained Decision Tree model is used to predict 'Menta_Health_Score'
- Predictions are made using the unseen test features('X_test')
- The predicted scores are stored in 'y_pred_tree'

### Decision Tree - Evaluation

```python
# Calculate evaluation metrics for Decision Tree

tree_mae = mean_absolute_error(y_test,y_pred_tree)

tree_rmse = np.sqrt(mean_squared_error(y_test,y_pred_tree))

tree_r2 = r2_score(y_test, y_pred_tree)
```

```python
# Store results in DataFrame
tree_results = pd.DataFrame({
    "Model": ["Decision Tree"],
    "MAE" :[tree_mae],
    "RMSE" :[tree_rmse],
    "R2 Score": [tree_r2]
})

tree_results
```

### Key Insights

- Decision Tree achieved an **MAE of 0.3228**
- The **RMSE** is **0.5158**
- The **R2** Score is **0.8510**
- The final model will be selected after comparing all four models

# 3. Random Forest Regression

- Random Forest is made up of many Decision Trees 
- Each tree makes its own prediction
- The predictions from all trees are combined to produce the final prediction
- It can capture complex and non-linear relationships in the data

### Random Forest Regression model - Creation

```python
# Import Random Forest Regressor 
from sklearn.ensemble import RandomForestRegressor

# Create the Random Forest model
random_forest_model = RandomForestRegressor(random_state=42)

# Train the Random Forest model
random_forest_model.fit(X_train, y_train)

print("Random Forest Model Trained Successfully!")

```

### Random Forest - Prediction

```python
# Make predictions using the trained Random Forest model
y_pred_rf = random_forest_model.predict(X_test)

# Display the first 10 predictions
y_pred_rf[:10]
```

- The trained Random Forest model is used to predict 'Mental_Health_Score'
- Predictions are made using the unseed features('X_test')
- The predicted scores are stored in 'y_pred_rf'

### Random Forest - Evaluation

```python
# Calculate evaluation metrics for Random Forest

rf_mae = mean_absolute_error(y_test, y_pred_rf)

rf_rmse = np.sqrt(mean_squared_error(y_test,y_pred_rf))

rf_r2 = r2_score(y_test, y_pred_rf)
```

```python
# Store results in a DataFrame
rf_results = pd.DataFrame({
    "Model" : ["Random Forest"],
    "MAE" : [rf_mae],
    "RMSE" : [rf_rmse],
    "R2 Score": [rf_r2]
})

rf_results
```

### Key Insights

- Random Forest achieved an **MAE of 0.277949**
- The **RMSE** is **0.38565**
- The **R2** is **0.91669**
- Random Forest currently performs better than Linear Regression and Decision Tree
- The final model will be selected after evaluating Gradient Boosting

# Gradient Boosting Regression 
- Gradient Boosting build multiple decision trees one after another 
- Each new tree tries to improve the errors made by previous trees 
- It can capture complex and non-linear relationships in the data 
- The final prediction combines the contributions of all the trees


### Gradient Boosting Regression Model - Creation 

```python
# Import Gradient Boosting Regressor
from sklearn.ensemble import GradientBoostingRegressor

# Create Gradient Boost Model
gradient_boosting_model = GradientBoostingRegressor(
    random_state=42
)

# Train Gradient Boost Model
gradient_boosting_model.fit(X_train,y_train)

print("Gradient Boosting model trained successfully!")
```

### Gradient Boost - Prediction

```python
# Make predictions using the Gradient Boosting model
y_pred_grad = gradient_boosting_model.predict(X_test)

# Display the first 10 predictions
y_pred_grad[:10]
```

### Gradient Boost - Evaluation

```python
# Calculate evaluation metrics for Gradient Boost

gb_mae = mean_absolute_error(y_test, y_pred_grad)

gb_rmse = np.sqrt(mean_squared_error(y_test,y_pred_grad))

gb_r2 = r2_score(y_test, y_pred_grad)
```

```python
gb_results = pd.DataFrame({
    "Model" : ["Gradient Boost"],
    "MAE" : [gb_mae],
    "RMSE" : [gb_rmse],
    "R2 Score" : [gb_r2]
})

gb_results
```

### Key Insights

- Gradient Boosting achieved an **MAE of 0.410558**
- The **RMSE** is **0.529632**
- The **R2 Score** is **0.84287**
- Gradient Boosting performs better than Linear Regression but worse than Decision Tree and Random Forest on the test data
- Among the four models evaluated so far, **Random Forest shows the besst overall performance!**

## Final Model Comparison

```python
# Combine the evaluation results of all four models
model_comparison = pd.concat([
    linear_results, tree_results, rf_results, gb_results
] , ignore_index = True)

# Display the final comparison
model_comparison.sort_values(ascending=False, by= "R2 Score").reset_index(drop=True)

# df.sort_values(ascending=False, by= "column_name") --> to sort values by specific column


# df.reset_index(drop=True) --> Discard the old, jumbled index and reset it to a clean 0, 1, 2... sequence
```

### Key Insights

- Four regression models were evaluated on the same test dataset
- **Random Forest achieved the lowest MAE (0.2779)** and **lowest RMSE (0.3857)**
- Random Forest also achieved the **highest R² Score (0.9167)**
- Therefore, **Random Forest is the best-performing model among the four models tested**
- We will use Random Forest as the candidate model for the next stage of model development
