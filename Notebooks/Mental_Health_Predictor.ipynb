# Student's Mental Health Score Prediction System 

This notebook demonstrates the final Mental Health Prediction System using the trained Random Forest model

The system takes a student's academic, lifestyle, social media usage, sleep, physical activity, and stress information as input

It then:
- Predicts the student's Mental Health Score
- Assigns a mental health category
- Analyzes important lifestyle factors
- Provides personalized insights

The trained model and preprocessing pipeline are loaded fron the project files and do predictions

## 1. Import Required Libraries

Import the libraries required to load the trained model, prepare the student input, and perform the prediction

```python
# Import Libraries

#helps Python find our project modules
import sys

#helps create reliable file paths
from pathlib import Path

#loads saved ML model and preprocessor
import joblib

#creates the student input as a DataFrame
import pandas as pd 
```

## 2. Set the Project Path

The notebook is strored inside the 'Notebooks' folder, while the trained model and preprocessing pipeline are stored isnide the 'Models' folder

therefore identify project root before loading files

```python
# Find the project root directory

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "Notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

# Define the Models folder
MODELS_DIR = PROJECT_ROOT / "models"

print("Project root:", PROJECT_ROOT)
print("Models folder:", MODELS_DIR)
```

## 3. Load the Trained Model and Preprocessor

The Random Forest Model was trained and tuned in the previous notebooks

The preprocessing pipeline was also saved separately

load both files here in final notebook to make predictions using same processing steps used during model training 

```python
# Load the saved preprocessing pipeline

preprocessor = joblib.load( MODELS_DIR / "preprocessor.pkl")

# Load the final trianed Random Forest model
model = joblib.load( MODELS_DIR / "final_random_forest.pkl")

print("Preprocessor loaded successfully.")
print("Random Forest model loaded successfully")
```

```python
# Check which columns saved preprocessor expects

print("Columns expected by the saved preprocessor:")
print(preprocessor.feature_names_in_.tolist())
```

## 4. Create Student Input

The prediction system accepts student's demographic, academic, lifestyle, social media usage, sleep, physical activity, and stress information

The input uses same feature names and structure used during model training

```python
# Create one sample student's information for testing the model

student_data = {
    "Age": 20,
    "Gender": "Male",
    "Country_Processed": "India",
    "Academic_Level": "Undergraduate",
    "Most_Used_Platform": "Instagram",
    "Purpose_Of_Use": "Entertainment",
    "Stress_Level": "Low",
    "Avg_Daily_Usage_Hours": 2.0,
    "Daily_Unlocks": 30,
    "Study_Hours": 5.0,
    "Physical_Activity_Hours": 1.5,
    "Sleep_Hours_Per_Night": 8.0
}

# Convert input dictionary into a DataFrame
student_df = pd.DataFrame([student_data])

student_df
```

## 5. Check Input Features

Before preprocessing, verify that student input contains all features required by the trained preprocessing pipeline

```python
# Display input columns
print("Input Columns:")
print(student_df.columns.tolist())

print("\nInput Shape:", student_df.shape)
```

## 6. Preprocess Student Data

The saved preprocessing pipeline performs the same transformations used during model training

Categorical are encoded and numerical features are scaled before data is passed to Random Forest Model

```python
# Transofrm the student input using the saved preprocessing pipeline
student_processed = preprocessor.transform(student_df)

print("Student data preprocessed successfully")
print("Processed Shape:", student_processed.shape)
```

## 7. Predict Mental Helath Score

The preprocessed student data is now passed  to the trained Random Forest Model

The model predicts a continuous Mental Health Score Based on the student's input characterstics

```python
# Get the feature names after preprocessing
feature_names = preprocessor.get_feature_names_out()

# Convert processed data into a DataFrame with feature names
student_processed_df = pd.DataFrame(
    student_processed,
    columns=feature_names
)

# Predict Mental Health Score
predicted_score = model.predict(student_processed_df)[0]

# Display predicted score
print(f"Predicted Mental Health Score: {predicted_score:.2f}")
```

## 8. Determine Mental Health Category

the predicted score is converted into a meaningful category using scoring rule implemented in the project

The categories are:
- Low
- Moderate
- Good
- Excellent / Stable

```python
# Add project root to Python's module search path
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

# Impot project's existing category function
from Backend.feature_analysis import get_score_category

# Determine category of predicted score
score_category = get_score_category(predicted_score)

print(f"Mental Health Category: {score_category}")
```

## 9. Analyze Student Features

Analyze important lifestyle factors to identify area that may need attention

```python
# Import project's feature analysis functions
from Backend.feature_analysis import(
    analyze_student_features,
    create_ai_feature_context
)

# Analyze student's important lifestyle factors
feature_analysis = analyze_student_features(student_data, predicted_score)

# Create a text context from analysis
ai_context = create_ai_feature_context(
    student_data, predicted_score, feature_analysis
)

print("Feature analysis completed successfully!")
```

## 10. Student Lifestyle Analysis

The analysis shows the student's stable areas and factors that may need attention

```python
# Display the fature analysis results
for factor, details in feature_analysis.items():
    print(f"\n{factor}")
    print("-" * len(factor))
    print(details)
```

## 11. Personalized Insights

Generate a simple summary based on the prediction and feature analysis.

```python
# Display the overall prediction
print(f"Mental Health Score : {predicted_score:.2f}")
print(f"Category             : {score_category}")

# Display factors that may need attention
if feature_analysis["needs_attention"]:
    print("\nNeeds Attention:")
    for item in feature_analysis["needs_attention"]:
        print(f"- {item['label']}: {item['message']}")

# Display factors that can be improved
if feature_analysis["watch"]:
    print("\nCan Improve:")
    for item in feature_analysis["watch"]:
        print(f"- {item['label']}: {item['message']}")

# Display stable factors
print("\nStable Areas:")
for item in feature_analysis["stable"]:
    print(f"- {item['label']}: {item['interpretation']}")
```
